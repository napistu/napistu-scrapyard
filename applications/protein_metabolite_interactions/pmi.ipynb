```python
import os

from napistu_torch.utils.base_utils import ensure_path
from napistu_binding.load.sdf import download_chebi_sdf, extract_molecules_from_sdf

DATA_DIR = os.path.expanduser("~/Desktop/DATA/structures")
CHEBI_FILENAME = "chebi_3_stars.sdf"
PROTEIN_STRUCTURES_DIR = ensure_path(os.path.join(DATA_DIR, "protein_structures"))
DIFFDOCK_RESULTS_DIR = ensure_path(os.path.join(DATA_DIR, "diffdock_results"))

chebi_path = os.path.join(DATA_DIR, CHEBI_FILENAME)
if not os.path.isdir(PROTEIN_STRUCTURES_DIR):
    os.makedirs(PROTEIN_STRUCTURES_DIR, exist_ok=True)
if not os.path.isdir(DIFFDOCK_RESULTS_DIR):
    os.makedirs(DIFFDOCK_RESULTS_DIR, exist_ok=True)
```

```python
from logging import getLogger

logger = getLogger(__name__)
logger.setLevel("INFO")
```

```python
from napistu_binding.load.nstructure import nStructure
from napistu_binding.load.nmol import nMol
from napistu_binding.load.diffdock import DiffDockRunner, DiffDockManager
```

```python
ldh = nStructure.from_alphafold("P00338")  # Human LDH-A
structure_path = os.path.join(PROTEIN_STRUCTURES_DIR, "1CRN.pdb")
ldh.to_pdb_file(structure_path)
```

```python
nmol = nMol.from_smiles("CC(=O)O")
```

```python
DiffDockManager(ldh, nmol, PROTEIN_STRUCTURES_DIR, DIFFDOCK_RESULTS_DIR).ensure_dock(backend="docker")
```

```python
import logging
from typing import Union, List, Optional, Dict, Any
from pathlib import Path
```

```python
DiffDockRunner(backend="docker").predict_poses(
    protein = Path(structure_path),
    ligand = "CC(=O)C(=O)[O-]",
    out_dir = os.path.join(DIFFDOCK_RESULTS_DIR, "1CRN_CC(=O)C(=O)[O-]"),
    num_poses=10,
    verbose=True
)
```

```python
# download the chebi sdf file
if not os.path.isfile(chebi_path):
    download_chebi_sdf(chebi_path)
```

```python
# ~20s
molecules = extract_molecules_from_sdf(chebi_path, remove_invalid = True, verbose = True, suppress_rdkit_errors = True)
smiles = [mol.to_isomeric_smiles(mol) for mol in molecules]
```
